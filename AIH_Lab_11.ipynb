```python

# Download Heart Failure Clinical Records Dataset

url = "https://raw.githubusercontent.com/omswa513/Heart-Failure/main/heart_failure_clinical_records_dataset.csv"

df = pd.read_csv(url)

print("Dataset downloaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python

# ==========================================
# CELL 2: DATASET SUMMARY & VISUALIZATION
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# 1. Dataset Summary
print("DATASET SUMMARY")
print("=" * 40)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

# 2. Missing Values Heatmap
plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cmap='viridis',
    cbar=False
)

plt.title("Missing Values Heatmap")
plt.xlabel("Features")
plt.ylabel("Patient Records")
plt.show()

# 3. Histograms of Numerical Features
df.hist(
    figsize=(15, 12),
    bins=20,
    edgecolor='black'
)

plt.suptitle(
    "Distribution of Heart Failure Clinical Features",
    fontsize=15
)

plt.tight_layout()
plt.show()

# 4. Boxplots for Outlier Detection
numerical_columns = df.select_dtypes(
    include=np.number
).columns.drop('DEATH_EVENT')

plt.figure(figsize=(16, 10))

for i, col in enumerate(numerical_columns, 1):

    plt.subplot(3, 4, i)

    sns.boxplot(x=df[col])

    plt.title(col)

plt.suptitle("Outlier Detection Using Boxplots")
plt.tight_layout()
plt.show()

# 5. Target Variable Distribution
plt.figure(figsize=(6, 4))

sns.countplot(
    x='DEATH_EVENT',
    data=df
)

plt.title("Death Event Distribution")
plt.xlabel("Death Event (0 = Survived, 1 = Death)")
plt.ylabel("Number of Patients")
plt.show()

print("\nCELL 2 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# CELL 3: DATA CLEANING & PREPROCESSING
# ==========================================

from sklearn.model_selection import train_test_split

# Create a copy of original dataset
df_clean = df.copy()

# 1. Correct invalid values
df_clean.loc[df_clean['age'] < 0, 'age'] = np.nan

# Positive-only clinical measurements
positive_columns = [
    'creatinine_phosphokinase',
    'ejection_fraction',
    'platelets',
    'serum_creatinine',
    'serum_sodium',
    'time'
]

for col in positive_columns:
    df_clean.loc[df_clean[col] <= 0, col] = np.nan

# Ejection fraction cannot exceed 100
df_clean.loc[
    df_clean['ejection_fraction'] > 100,
    'ejection_fraction'
] = np.nan

# Binary categorical columns
categorical_columns = [
    'anaemia',
    'diabetes',
    'high_blood_pressure',
    'sex',
    'smoking'
]

for col in categorical_columns:
    df_clean.loc[
        ~df_clean[col].isin([0, 1]),
        col
    ] = np.nan

# Remove rows with invalid target values
df_clean = df_clean[
    df_clean['DEATH_EVENT'].isin([0, 1])
].copy()

print("Invalid values corrected!")

# 2. Separate features and target
X = df_clean.drop('DEATH_EVENT', axis=1)
y = df_clean['DEATH_EVENT'].astype(int)

# 3. Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)

# 4. Identify numerical features
numeric_columns = X_train.select_dtypes(
    include=np.number
).columns.tolist()

# 5. Missing value imputation
# Numerical features: Median
for col in numeric_columns:

    median_value = X_train[col].median()

    X_train[col] = X_train[col].fillna(median_value)
    X_test[col] = X_test[col].fillna(median_value)

# Categorical features: Mode
for col in categorical_columns:

    mode_value = X_train[col].mode()[0]

    X_train[col] = X_train[col].fillna(mode_value)
    X_test[col] = X_test[col].fillna(mode_value)

print("\nMissing values imputed successfully!")

# 6. Outlier detection and capping using IQR
outlier_columns = [
    col for col in numeric_columns
    if col not in categorical_columns
]

outlier_limits = {}

for col in outlier_columns:

    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_limits[col] = (
        lower_limit,
        upper_limit
    )

    X_train[col] = X_train[col].clip(
        lower_limit,
        upper_limit
    )

    X_test[col] = X_test[col].clip(
        lower_limit,
        upper_limit
    )

print("Outlier treatment completed!")

# 7. Final verification
print("\nFINAL DATA CLEANING REPORT")
print("=" * 40)

print("Training Missing Values:",
      X_train.isnull().sum().sum())

print("Testing Missing Values:",
      X_test.isnull().sum().sum())

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

print("\nTarget Distribution:")
print(y.value_counts())

print("\nCELL 3 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# CELL 4: FEATURE ENGINEERING & NORMALIZATION
# ==========================================

from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# 1. Create new features
def create_features(data, creatinine_min, creatinine_max):

    data = data.copy()

    # Create age groups
    data['age_group'] = pd.cut(
        data['age'],
        bins=[-np.inf, 40, 60, np.inf],
        labels=['Young', 'Middle-aged', 'Senior'],
        right=False
    )

    # Normalize ejection fraction risk
    ef_risk = 1 - (data['ejection_fraction'] / 100)

    # Normalize serum creatinine using training limits
    if creatinine_max > creatinine_min:
        creatinine_risk = (
            (data['serum_creatinine'] - creatinine_min) /
            (creatinine_max - creatinine_min)
        )
    else:
        creatinine_risk = 0

    # Calculate combined risk score
    data['risk_score'] = ef_risk * creatinine_risk

    return data


# Calculate limits from training data
creatinine_min = X_train['serum_creatinine'].min()
creatinine_max = X_train['serum_creatinine'].max()

# Apply feature engineering
X_train_fe = create_features(
    X_train,
    creatinine_min,
    creatinine_max
)

X_test_fe = create_features(
    X_test,
    creatinine_min,
    creatinine_max
)

# Display newly created features
print("FEATURE ENGINEERING RESULTS:")
display(
    X_train_fe[
        ['age', 'age_group',
         'ejection_fraction',
         'serum_creatinine',
         'risk_score']
    ].head()
)

# 2. Identify categorical and numerical features
categorical_features = categorical_columns + ['age_group']

numerical_features = [
    col for col in X_train_fe.columns
    if col not in categorical_features
]

print("\nNumerical Features:", numerical_features)
print("\nCategorical Features:", categorical_features)

# 3. One-Hot Encoding
try:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse=False
    )

# 4. Apply MinMaxScaler and OneHotEncoder
preprocessor = ColumnTransformer(
    transformers=[
        ('num', MinMaxScaler(), numerical_features),
        ('cat', encoder, categorical_features)
    ]
)

# Fit on training data only
X_train_processed = preprocessor.fit_transform(X_train_fe)

# Transform test data
X_test_processed = preprocessor.transform(X_test_fe)

# 5. Get feature names
feature_names = preprocessor.get_feature_names_out()

# Convert into DataFrames
X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

# Display final processed data
print("\nPROCESSED TRAINING DATA:")
display(X_train_processed.head())

print("\nPROCESSED TEST DATA:")
display(X_test_processed.head())

print("\nFinal Training Shape:", X_train_processed.shape)
print("Final Testing Shape:", X_test_processed.shape)

print("\nMinimum Feature Value:",
      X_train_processed.min().min())

print("Maximum Feature Value:",
      X_train_processed.max().max())

print("\nCELL 4 COMPLETED SUCCESSFULLY!")
```

```python

# CELL 5: MODEL TRAINING AND EVALUATION

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report,
    roc_auc_score, RocCurveDisplay
)
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Create and train the Random Forest model
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=5,
    min_samples_split=5,
    random_state=42,
    class_weight='balanced'
)

model.fit(X_train_processed, y_train)

# 2. Make predictions
y_pred = model.predict(X_test_processed)
y_prob = model.predict_proba(X_test_processed)[:, 1]

# 3. Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_prob)

print("MODEL PERFORMANCE")
print("-------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")

# 4. Classification report
print("\nClassification Report:")
print(classification_report(
    y_test, y_pred,
    target_names=["Survived (0)", "Death Event (1)"],
    zero_division=0
))

# 5. Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=["Survived", "Death Event"],
    yticklabels=["Survived", "Death Event"]
)
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")
plt.show()

# 6. ROC Curve
RocCurveDisplay.from_predictions(y_test, y_prob)
plt.title("ROC Curve - Random Forest")
plt.show()

# 7. Feature Importance
importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
}).sort_values(by="Importance", ascending=False)

print("\nTop 10 Important Features:")
display(importance.head(10))
```
